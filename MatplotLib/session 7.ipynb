task 1

Use plt.subplot() to create a figure with 2 rows and 1 column of plots showing your daily step count and water intake (in litres) for the last 7 days. Plot both as line charts, one in each subplot.

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [5000, 7000, 6500, 8000, 9000, 7500, 10000]
water = [2, 2.5, 2, 3, 2.5, 3, 2.5]

# First subplot
plt.subplot(2, 1, 1)

plt.plot(days, steps, marker="o")
plt.title("Daily Step Count")
plt.ylabel("Steps")
plt.grid()

# Second subplot
plt.subplot(2, 1, 2)

plt.plot(days, water, marker="o")
plt.title("Daily Water Intake")
plt.xlabel("Days")
plt.ylabel("Water (Litres)")
plt.grid()

plt.tight_layout()
plt.show()
```

task 2

Create a 2x2 multi-panel figure using plt.subplots(). In each subplot, display the number of Instagram posts you made in each quarter (Q1, Q2, Q3, Q4) for the past year, using different chart types: line, bar, scatter, and pie.

```python
import matplotlib.pyplot as plt

quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [15, 22, 18, 30]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# 1. Line chart
axes[0, 0].plot(quarters, posts, marker="o")
axes[0, 0].set_title("Line Chart")
axes[0, 0].set_xlabel("Quarter")
axes[0, 0].set_ylabel("Posts")
axes[0, 0].grid()

# 2. Bar chart
axes[0, 1].bar(quarters, posts)
axes[0, 1].set_title("Bar Chart")
axes[0, 1].set_xlabel("Quarter")
axes[0, 1].set_ylabel("Posts")

# 3. Scatter plot
axes[1, 0].scatter(quarters, posts)
axes[1, 0].set_title("Scatter Plot")
axes[1, 0].set_xlabel("Quarter")
axes[1, 0].set_ylabel("Posts")
axes[1, 0].grid()

# 4. Pie chart
axes[1, 1].pie(posts, labels=quarters, autopct="%1.1f%%")
axes[1, 1].set_title("Pie Chart")

fig.suptitle("Instagram Posts by Quarter")

plt.tight_layout()
plt.show()
```

task 3


Make a grid of 1 row and 3 columns using plt.subplots(), and plot the number of food orders you placed on Zomato, Swiggy, and Domino's over the last 6 months, one app per subplot. Share the y-axis among all subplots so you can easily compare order counts across apps.<br><br><em><strong>Hint:</strong> Use the 'sharey' parameter in plt.subplots().</em>

```python
import matplotlib.pyplot as plt

months = ["Mar", "Apr", "May", "Jun", "Jul", "Aug"]

zomato = [8, 10, 12, 9, 14, 16]
swiggy = [6, 9, 10, 11, 13, 15]
dominos = [4, 5, 7, 6, 8, 10]

fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharey=True)

# Zomato
axes[0].plot(months, zomato, marker="o")
axes[0].set_title("Zomato")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Number of Orders")
axes[0].grid()

# Swiggy
axes[1].plot(months, swiggy, marker="o")
axes[1].set_title("Swiggy")
axes[1].set_xlabel("Month")
axes[1].grid()

# Domino's
axes[2].plot(months, dominos, marker="o")
axes[2].set_title("Domino's")
axes[2].set_xlabel("Month")
axes[2].grid()

fig.suptitle("Food Orders Comparison")

plt.tight_layout()
plt.show()
```

task 4


Arrange a 2x2 grid of subplots using plt.subplots() to visualize your monthly mobile data usage (in GB) for four different apps: YouTube, Instagram, Spotify, and WhatsApp. Ensure all subplots share the same x-axis (months) for easy comparison.<br><br><em><strong>Hint:</strong> Use the 'sharex' parameter in plt.subplots().</em>

```python
import matplotlib.pyplot as plt

months = ["Mar", "Apr", "May", "Jun", "Jul", "Aug"]

youtube = [20, 25, 22, 30, 28, 35]
instagram = [12, 15, 14, 18, 20, 22]
spotify = [15, 18, 17, 21, 24, 26]
whatsapp = [10, 12, 11, 14, 15, 17]

fig, axes = plt.subplots(
    2, 2,
    figsize=(10, 8),
    sharex=True
)

# YouTube
axes[0, 0].plot(months, youtube, marker="o")
axes[0, 0].set_title("YouTube")
axes[0, 0].set_ylabel("Data Used (GB)")
axes[0, 0].grid()

# Instagram
axes[0, 1].plot(months, instagram, marker="o")
axes[0, 1].set_title("Instagram")
axes[0, 1].set_ylabel("Data Used (GB)")
axes[0, 1].grid()

# Spotify
axes[1, 0].plot(months, spotify, marker="o")
axes[1, 0].set_title("Spotify")
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("Data Used (GB)")
axes[1, 0].grid()

# WhatsApp
axes[1, 1].plot(months, whatsapp, marker="o")
axes[1, 1].set_title("WhatsApp")
axes[1, 1].set_xlabel("Month")
axes[1, 1].set_ylabel("Data Used (GB)")
axes[1, 1].grid()

fig.suptitle("Monthly Mobile Data Usage")

plt.tight_layout()
plt.show()
```
