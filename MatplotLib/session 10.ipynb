task 1

Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.


```python
import matplotlib.pyplot as plt

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

users = [
    1200, 1350, 1280, 1450, 1600, 1550,
    1700, 1850, 1800, 1950, 2100, 2000
]

plt.figure(figsize=(10, 5))

plt.plot(months, users, marker="o")

# Find highest value
max_users = max(users)
max_index = users.index(max_users)

# Annotate highest month
plt.annotate(
    f"Highest: {max_users}",
    xy=(months[max_index], max_users),
    xytext=(months[max_index], max_users + 200),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Active Users")
plt.xlabel("Month")
plt.ylabel("Active Users")

plt.grid()
plt.show()
```

task 2

Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>

```python
import matplotlib.pyplot as plt

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

orders = [
    1200, 1350, 1100, 1450, 1600, 1500,
    1300, 1250, 1050, 1400, 1550, 1700
]

plt.figure(figsize=(10, 5))

plt.bar(months, orders)

# Find lowest value
min_orders = min(orders)
min_index = orders.index(min_orders)

# Add arrow annotation
plt.annotate(
    f"Lowest: {min_orders}",
    xy=(min_index, min_orders),
    xytext=(min_index - 1, min_orders + 400),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Zomato Orders")
plt.xlabel("Month")
plt.ylabel("Number of Orders")

plt.grid(axis="y")

plt.show()
```

task 3

Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.

```python
import matplotlib.pyplot as plt

payment_methods = [
    "UPI",
    "Card",
    "Wallet",
    "Cash"
]

usage = [55, 25, 15, 5]

plt.figure(figsize=(8, 6))

plt.pie(
    usage,
    autopct="%1.1f%%"
)

plt.title("E-Commerce Payment Methods")

plt.legend(
    payment_methods,
    loc="lower right"
)

plt.show()
```

task 4

Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>

```python
import matplotlib.pyplot as plt

movies = [
    "Movie A", "Movie B", "Movie C", "Movie D",
    "Movie E", "Movie F", "Movie G", "Movie H"
]

ratings = [7.2, 8.1, 7.5, 9.2, 8.5, 7.8, 8.9, 7.0]

reviews = [1200, 2500, 1800, 4200, 3000, 2100, 3500, 1000]

plt.figure(figsize=(10, 5))

plt.scatter(reviews, ratings)

# Find highest rating
max_rating = max(ratings)
max_index = ratings.index(max_rating)

# Add text box
plt.annotate(
    f"Highest Rating\n{movies[max_index]}: {max_rating}",
    xy=(reviews[max_index], max_rating),
    xytext=(reviews[max_index] + 500, max_rating - 0.5),
    bbox=dict(
        boxstyle="round",
        facecolor="white",
        edgecolor="black"
    ),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Movie Ratings vs Number of Reviews")
plt.xlabel("Number of Reviews")
plt.ylabel("Rating")

plt.grid()
plt.show()
```
