task 1

Create a stacked bar chart using Matplotlib that shows the number of food orders from Zomato and Swiggy for each day of the week. Use different colors for each platform and add a legend.

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato = [120, 150, 140, 180, 200, 220, 210]
swiggy = [100, 130, 120, 150, 180, 190, 200]

plt.bar(days, zomato, label="Zomato")
plt.bar(days, swiggy, bottom=zomato, label="Swiggy")

plt.title("Daily Food Orders")
plt.xlabel("Days")
plt.ylabel("Number of Orders")

plt.legend()
plt.show()
```

task 2

Plot an area chart that visualizes the daily active users on Spotify and YouTube over a 7-day period using Matplotlib's fill_between(). Make sure each platform's area is filled with a different color and add labels for clarity.

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify = [1200, 1350, 1400, 1500, 1600, 1800, 1750]
youtube = [1000, 1200, 1300, 1450, 1500, 1700, 1650]

plt.figure(figsize=(10, 5))

plt.plot(days, spotify, label="Spotify")
plt.fill_between(days, spotify, alpha=0.4)

plt.plot(days, youtube, label="YouTube")
plt.fill_between(days, youtube, alpha=0.4)

plt.title("Spotify vs YouTube Daily Active Users")
plt.xlabel("Days")
plt.ylabel("Daily Active Users")

plt.legend()
plt.show()
```

task 3

Simulate the average delivery time (in minutes) for Flipkart orders over 10 days, and add error bars to represent the standard deviation for each day's delivery time using Matplotlib's errorbar() function.<br><br><em><strong>Hint:</strong> Use the yerr parameter to show error bars for each data point.</em>

```python
import matplotlib.pyplot as plt

days = [
    "Day 1", "Day 2", "Day 3", "Day 4", "Day 5",
    "Day 6", "Day 7", "Day 8", "Day 9", "Day 10"
]

average_time = [35, 38, 32, 40, 36, 42, 39, 34, 37, 41]

std_dev = [3, 4, 2, 5, 3, 4, 3, 2, 4, 5]

plt.figure(figsize=(10, 5))

plt.errorbar(
    days,
    average_time,
    yerr=std_dev,
    fmt="o-",
    capsize=5
)

plt.title("Flipkart Average Delivery Time")
plt.xlabel("Days")
plt.ylabel("Delivery Time (Minutes)")

plt.grid()
plt.show()
```

task 4

Plot a time series line chart showing your daily Paytm wallet balance over the last 14 days using Matplotlib. Format the x-axis to display dates in 'DD MMM' format (e.g., '12 Jun').

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime, timedelta

start_date = datetime(2026, 8, 16)

dates = [
    start_date + timedelta(days=i)
    for i in range(14)
]

balance = [
    5000, 4700, 4900, 4500,
    5200, 5100, 4800, 5500,
    5300, 5700, 5400, 6000,
    5800, 6200
]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(dates, balance, marker="o")

ax.set_title("Paytm Wallet Balance - Last 14 Days")
ax.set_xlabel("Date")
ax.set_ylabel("Wallet Balance (₹)")

# Format dates as DD MMM
ax.xaxis.set_major_formatter(
    mdates.DateFormatter("%d %b")
)

plt.xticks(rotation=45)

ax.grid(True)

plt.tight_layout()
plt.show()
```
