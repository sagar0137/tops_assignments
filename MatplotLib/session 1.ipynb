task 1

Install Matplotlib in your Python environment and write a script named hello_plot.py that creates a simple line plot showing the number of daily steps you walked over the past 7 days

```python
import matplotlib.pyplot as plt

# Data for 7 days
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [6500, 7200, 8000, 6000, 9000, 7500, 8200]

# Create line plot
plt.plot(days, steps, marker="o")

# Add labels and title
plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.title("Daily Steps Over the Past 7 Days")

# Display grid
plt.grid(True)

# Display plot
plt.show()
```

task 3


Write a Python script that creates a Matplotlib figure with two Axes: one showing your daily Spotify listening hours for a week and the other showing your daily YouTube viewing hours for the same week. Display both plots in a single figure.

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_hours = [2, 3, 1.5, 2.5, 3, 4, 2]
youtube_hours = [1, 2, 2.5, 1.5, 2, 3, 2.5]

# Create one figure with two Axes
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Spotify plot
ax[0].plot(days, spotify_hours, marker="o")
ax[0].set_xlabel("Days")
ax[0].set_ylabel("Listening Hours")
ax[0].set_title("Daily Spotify Listening")

# YouTube plot
ax[1].plot(days, youtube_hours, marker="o")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Viewing Hours")
ax[1].set_title("Daily YouTube Viewing")

plt.tight_layout()

plt.show()
```

task 4

Add axis labels, a title, and a grid to your plot from the previous task to clearly communicate the information, mimicking the style of analytics dashboards in apps like Instagram Insights.<br><br><em><strong>Hint:</strong> Use set_xlabel(), set_ylabel(), set_title(), and grid() methods on the Axes object.</em>

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_hours = [2, 3, 1.5, 2.5, 3, 4, 2]
youtube_hours = [1, 2, 2.5, 1.5, 2, 3, 2.5]

# Create figure and two Axes
fig, ax = plt.subplots(1, 2, figsize=(12, 5))


# Spotify Axes

ax[0].plot(days, spotify_hours, marker="o")

ax[0].set_xlabel("Days")
ax[0].set_ylabel("Hours")
ax[0].set_title("Spotify Listening Hours")
ax[0].grid(True)

# YouTube Axes

ax[1].plot(days, youtube_hours, marker="o")

ax[1].set_xlabel("Days")
ax[1].set_ylabel("Hours")
ax[1].set_title("YouTube Viewing Hours")
ax[1].grid(True)

# Overall figure title
fig.suptitle("Weekly Entertainment Analytics")

plt.tight_layout()

plt.show()
```
