Load a CSV file containing Flipkart product ratings and prices, then use Seaborn's jointplot with kind='scatter' to visualize the relationship between rating and price

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("flipkart_products.csv")

sns.jointplot(
    x="rating",
    y="price",
    data=df,
    kind="scatter"
)

plt.show()
```

task 4

Compare the visual insights you get from scatter, hexbin, and KDE jointplots using the same dataset of Zomato restaurant reviews (rating vs number of reviews). Briefly describe which plot type helps you understand the data best and why.


```python
data = {
    "rating": [
        3.1, 3.2, 3.4, 3.5, 3.6,
        3.7, 3.8, 3.9, 4.0, 4.1,
        4.2, 4.3, 4.4, 4.5, 4.6,
        3.8, 4.0, 4.2, 4.3, 4.5,
        3.5, 3.7, 3.9, 4.1, 4.4,
        4.6, 4.7, 3.3, 3.6, 4.0
    ],
    
    "reviews": [
        50, 80, 100, 120, 150,
        180, 220, 250, 300, 350,
        400, 450, 500, 600, 700,
        200, 280, 380, 480, 550,
        130, 170, 240, 330, 520,
        650, 750, 90, 140, 310
    ]
}

df = pd.DataFrame(data)
```

scatter jointplot

```python
sns.jointplot(
    x="rating",
    y="reviews",
    data=df,
    kind="scatter"
)

plt.show()
```

Hexbin Jointplot

```python
sns.jointplot(
    x="rating",
    y="reviews",
    data=df,
    kind="hex"
)

plt.show()
```

KDE Jointplot

```python
sns.jointplot(
    x="rating",
    y="reviews",
    data=df,
    kind="kde"
)

plt.show()
```

Scatterplot is best when the dataset is small and you want to inspect individual restaurants and possible outliers.

Hexbin is better when the dataset is large and many points overlap because it shows where observations are concentrated.

KDE is best when your main goal is to understand the overall density and distribution of the two variables.
