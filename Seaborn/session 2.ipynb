task 1

Use sns.countplot() to visualize the number of orders placed by each payment method (e.g., UPI, Card, Cash) in a sample Swiggy orders DataFrame with at least 3 payment types.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Swiggy orders data
data = {
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "payment_method": [
        "UPI", "Card", "UPI", "Cash", "UPI",
        "Card", "Cash", "UPI", "Card", "UPI"
    ]
}

df = pd.DataFrame(data)

```

```python
sns.countplot(x="payment_method", data=df)

plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.title("Number of Orders by Payment Method")

plt.show()
```

task 2

Create a barplot using sns.barplot() to show the average delivery time (in minutes) for each food category (e.g., Pizza, Burger, Biryani, South Indian) from a Zomato-style dataset.<br><br><em><strong>Hint:</strong> Use the estimator parameter to show the mean delivery time for each category.</em>

```python

data = {
    "food_category": [
        "Pizza", "Pizza", "Pizza",
        "Burger", "Burger", "Burger",
        "Biryani", "Biryani", "Biryani",
        "South Indian", "South Indian", "South Indian"
    ],
    
    "delivery_time": [
        30, 35, 32,
        25, 28, 30,
        40, 45, 42,
        20, 25, 22
    ]
}

df = pd.DataFrame(data)

sns.barplot(
    x="food_category",
    y="delivery_time",
    data=df,
    estimator="mean"
)

plt.xlabel("Food Category")
plt.ylabel("Average Delivery Time (Minutes)")
plt.title("Average Delivery Time by Food Category")

plt.show()
```

task 3

Modify your barplot to display the total number of orders instead of the average by setting estimator to sum in sns.barplot().

```python

data = {
    "food_category": [
        "Pizza", "Pizza", "Pizza",
        "Burger", "Burger", "Burger",
        "Biryani", "Biryani", "Biryani",
        "South Indian", "South Indian", "South Indian"
    ],
    
    "order_count": [
        10, 15, 12,
        20, 18, 22,
        8, 10, 12,
        15, 18, 20
    ]
}

df = pd.DataFrame(data)

sns.barplot(
    x="food_category",
    y="order_count",
    data=df,
    estimator="sum"
)

plt.xlabel("Food Category")
plt.ylabel("Total Number of Orders")
plt.title("Total Orders by Food Category")

plt.show()
```

Plot a countplot of movie genres from a BookMyShow-style dataset, but remove the confidence interval from the plot.<br><br><em><strong>Hint:</strong> Use the ci=None parameter to remove confidence intervals in sns.barplot().</em>

```python

data = {
    "movie_name": [
        "Movie A", "Movie B", "Movie C", "Movie D",
        "Movie E", "Movie F", "Movie G", "Movie H",
        "Movie I", "Movie J"
    ],
    
    "genre": [
        "Action", "Comedy", "Action", "Drama",
        "Comedy", "Action", "Romance", "Drama",
        "Action", "Comedy"
    ]
}

df = pd.DataFrame(data)

sns.countplot(
    x="genre",
    data=df
)

plt.xlabel("Movie Genre")
plt.ylabel("Number of Movies")
plt.title("Number of Movies by Genre")

plt.show()
```
