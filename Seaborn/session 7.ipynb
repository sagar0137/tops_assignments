task 1

Create a Pandas DataFrame with at least 5 columns representing different features of food delivery orders (e.g., delivery_time, order_value, tip_amount, rating, delivery_distance) and 20 rows of random data, then calculate and display the correlation matrix using the .corr() function.


```python
import pandas as pd
import numpy as np

# Create random food delivery data
np.random.seed(42)

df = pd.DataFrame({
    "delivery_time": np.random.randint(20, 60, 20),
    "order_value": np.random.randint(200, 1500, 20),
    "tip_amount": np.random.randint(10, 200, 20),
    "rating": np.round(np.random.uniform(3.0, 5.0, 20), 1),
    "delivery_distance": np.round(np.random.uniform(1, 15, 20), 2)
})

correlation_matrix = df.corr()

correlation_matrix
```

task 2

Using Seaborn, plot a heatmap of the correlation matrix you generated for the food delivery data, and add annotations to show the correlation values inside each cell.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(
    correlation_matrix,
    annot=True
)

plt.title("Food Delivery Correlation Heatmap")
plt.show()
```

task 3

Change the color palette of your Seaborn heatmap to a diverging palette (such as 'coolwarm' or 'RdYlGn') to better visualize positive and negative correlations, and explain which two features in your data have the strongest negative correlation.<br><br><em><strong>Hint:</strong> Use the cmap parameter in sns.heatmap().</em>

```python
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Food Delivery Correlation Heatmap")
plt.show()
```

```python
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="RdYlGn"
)

plt.title("Correlation Heatmap - RdYlGn")
plt.show()
```

task 4

Take the Flipkart mobile phones dataset (or any small CSV with product features you can find online), compute the correlation matrix for at least 4 numeric columns (e.g., price, rating, RAM, battery_capacity), and generate a Seaborn heatmap with annotations and a diverging color palette.

```python

data = {
    "price": [
        12000, 15000, 18000, 22000, 25000,
        28000, 32000, 35000, 40000, 45000,
        14000, 17000, 20000, 24000, 27000
    ],

    "rating": [
        3.8, 4.0, 4.1, 4.2, 4.3,
        4.4, 4.5, 4.3, 4.6, 4.5,
        3.9, 4.1, 4.2, 4.3, 4.4
    ],

    "RAM": [
        4, 4, 6, 6, 8,
        8, 8, 12, 12, 12,
        4, 6, 6, 8, 8
    ],

    "battery_capacity": [
        4000, 4500, 5000, 5000, 5200,
        5500, 5000, 6000, 5500, 6000,
        4200, 4800, 5000, 5200, 5400
    ]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

print(correlation_matrix)
```

```python
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Flipkart Mobile Phone Correlation Heatmap")
plt.show()
```
