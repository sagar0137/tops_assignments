task 1

Use sns.set() to apply the 'poster' context to a Seaborn barplot showing the number of songs streamed per day on Spotify, and observe how the plot elements change compared to the default context.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "day": [
        "Mon", "Tue", "Wed", "Thu",
        "Fri", "Sat", "Sun"
    ],
    
    "songs_streamed": [
        45, 60, 52, 70, 85, 100, 90
    ]
}

df = pd.DataFrame(data)


sns.set(context="poster")

sns.barplot(
    x="day",
    y="songs_streamed",
    data=df
)

plt.xlabel("Day")
plt.ylabel("Songs Streamed")
plt.title("Daily Spotify Songs Streamed")

plt.show()
```

task 3

Generate a Seaborn lineplot of daily step counts (simulate data for 2 weeks) and style it using the 'muted' palette and 'notebook' context for a clean, presentation-ready look.

```python

data = {
    "day": [
        "Day 1", "Day 2", "Day 3", "Day 4",
        "Day 5", "Day 6", "Day 7",
        "Day 8", "Day 9", "Day 10", "Day 11",
        "Day 12", "Day 13", "Day 14"
    ],

    "steps": [
        6500, 7200, 8000, 6800,
        9000, 9500, 7500,
        8200, 7800, 8800, 9200,
        8500, 10000, 9000
    ]
}

df = pd.DataFrame(data)

sns.set(
    context="notebook",
    palette="muted"
)

sns.lineplot(
    x="day",
    y="steps",
    data=df,
    marker="o"
)

plt.xlabel("Day")
plt.ylabel("Daily Steps")
plt.title("Daily Step Counts - 2 Weeks")

plt.xticks(rotation=45)

plt.show()
```

task 4

Refactor an existing Seaborn plot by switching its palette from 'deep' to 'colorblind' and changing the context from 'talk' to 'poster'. Briefly describe the visual differences you notice after applying these changes.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample dataset
data = {
    "category": ["Electronics", "Clothing", "Home Decor", "Books", "Toys"],
    "sales": [15000, 8500, 9200, 4300, 6100]
}

df = pd.DataFrame(data)

# Original theme settings
sns.set_theme(
    context="talk",
    palette="deep"
)

sns.barplot(
    x="category",
    y="sales",
    hue="category",
    legend=False,
    data=df
)

plt.title("Product Sales (Talk + Deep)")
plt.xlabel("Category")
plt.ylabel("Sales ($)")
plt.show()
```

```python

# Refactored theme settings
sns.set_theme(
    context="poster",
    palette="colorblind"
)

sns.barplot(
    x="category",
    y="sales",
    hue="category",
    legend=False,
    data=df
)

plt.title("Product Sales (Poster + Colorblind)")
plt.xlabel("Category")
plt.ylabel("Sales ($)")
plt.show()
```

task 5

Design a Seaborn theme that mimics the look of a Flipkart sales dashboard: use sns.set() to set a 'deep' palette, 'talk' context, and customize the font scale and grid style. Apply your theme to a barplot showing sales of 5 product categories.

```python

data = {
    "category": [
        "Mobiles",
        "Laptops",
        "TVs",
        "Headphones",
        "Accessories"
    ],

    "sales": [
        850000,
        650000,
        500000,
        300000,
        400000
    ]
}

df = pd.DataFrame(data)

sns.set(
    context="talk",
    style="whitegrid",
    palette="deep",
    font_scale=1.0
)

sns.barplot(
    x="category",
    y="sales",
    data=df
)

plt.xlabel("Product Category")
plt.ylabel("Sales (₹)")
plt.title("Flipkart Sales Dashboard")

plt.xticks(rotation=20)

plt.show()
```
