task 1

Load a CSV file containing Flipkart product prices and use sns.boxplot() to visualize the price distribution

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("flipkart_products.csv")

sns.boxplot(
    y=df["price"]
)

plt.ylabel("Product Price (₹)")
plt.title("Flipkart Product Price Distribution")

plt.show()
```

task 2

Use sns.violinplot() to compare the order amounts for two different Zomato restaurant categories (e.g., 'Fast Food' vs 'Desserts') from a sample dataset

```python
data = {
    "category": [
        "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Desserts", "Desserts", "Desserts", "Desserts",
        "Desserts", "Desserts", "Desserts", "Desserts"
    ],
    
    "order_amount": [
        250, 300, 350, 400,
        280, 450, 320, 380,
        150, 180, 220, 250,
        200, 300, 170, 280
    ]
}

df = pd.DataFrame(data)

sns.violinplot(
    x="category",
    y="order_amount",
    data=df
)

plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")
plt.title("Order Amount Distribution by Restaurant Category")

plt.show()
```

Create a violin plot using sns.violinplot() to show the distribution of daily step counts collected from a fitness app like Google Fit, and explain what the shape of the plot tells you about user activity.

```python
steps = [
    4500, 5200, 6000, 6500, 7000,
    7200, 7500, 7800, 8000, 8200,
    8500, 8700, 9000, 9200, 9500,
    9800, 10000, 10500, 11000, 11500,
    5500, 6300, 6800, 7300, 7600,
    8100, 8600, 9000, 9400, 10200
]

df = pd.DataFrame({
    "daily_steps": steps
})

sns.violinplot(
    y=df["daily_steps"]
)

plt.ylabel("Daily Steps")
plt.title("Distribution of Daily Step Counts")

plt.show()
```
