task 1

Import Seaborn and use sns.histplot() to visualize the distribution of delivery times (in minutes) from a sample Swiggy order dataset with at least 30 values.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

delivery_times = [
    25, 30, 35, 28, 32, 40, 45, 30, 27, 35,
    38, 42, 25, 29, 31, 36, 40, 33, 28, 34,
    45, 50, 32, 30, 27, 39, 41, 35, 29, 33,
    37, 43, 31, 26, 34
]

df = pd.DataFrame({
    "delivery_time": delivery_times
})

sns.histplot(
    data=df,
    x="delivery_time"
)

plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")
plt.title("Distribution of Swiggy Delivery Times")

plt.show()
```

task 2

Download or create a list of daily step counts (e.g., from a fitness app) for one month and plot the distribution using sns.kdeplot(), showing the probability density curve.

```python
steps = [
    6500, 7200, 8000, 6000, 9000,
    7500, 8200, 6800, 7100, 8500,
    9200, 7800, 6900, 7300, 8800,
    9500, 7600, 8100, 6700, 7000,
    8400, 9100, 7900, 8600, 7400,
    9800, 6300, 7700, 8300, 8900
]

df = pd.DataFrame({
    "steps": steps
})

sns.kdeplot(
    data=df,
    x="steps"
)

plt.xlabel("Daily Steps")
plt.ylabel("Probability Density")
plt.title("Distribution of Daily Step Counts")

plt.show()
```

task 3

Combine sns.histplot() and sns.kdeplot() to visualize the distribution of monthly spending amounts (in rupees) on Zomato orders, using the same dataset for both plots on a single figure.<br><br><em><strong>Hint:</strong> Use the same data and overlay the KDE curve on the histogram by calling both functions on the same axes.</em>

```python
spending = [
    450, 700, 1200, 850, 1500,
    600, 950, 1800, 1100, 750,
    1300, 900, 2000, 650, 1000,
    1250, 1700, 800, 1400, 2200,
    550, 1050, 1600, 1350, 1900,
    720, 1150, 1550, 980, 1450
]

df = pd.DataFrame({
    "monthly_spending": spending
})

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="monthly_spending",
    kde=True
)

plt.xlabel("Monthly Zomato Spending (₹)")
plt.ylabel("Number of Customers")
plt.title("Distribution of Monthly Zomato Spending")

plt.show()
```
