```python

#TASK 1--  Download the Kaggle Bank Marketing Dataset and use Pandas to load it into a DataFrame named bank_df, then display the first 10 rows and the DataFrame's info().

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd



bank_df = pd.read_csv("bank-additional-full.csv", sep=";")

print(bank_df.head(10))
print(bank_df.info())
```

```python
#TASK 2-- Use Seaborn's countplot to visualize how many customers in the dataset subscribed to a term deposit (column 'y'), and display the plot with appropriate axis labels.
bank_df = pd.read_csv("bank-additional-full.csv", sep=";")
 
sns.countplot(x="y", data=bank_df)
plt.xlabel("Subscribed to Term Deposit")
plt.ylabel("Number of Customers")
plt.title("Term Deposit Subscription Count")
plt.savefig("bank_countplot.png", dpi=150)
plt.show()
 
```

```python
#TASK 3-- Analyze the relationship between customer age and term deposit subscription by creating a Seaborn boxplot with 'y' on the x-axis and 'age' on the y-axis.<br><br><em><strong>Hint:</strong> This helps you see if certain age groups are more likely to subscribe.</em>
sns.boxplot(x="y", y="age", data=bank_df)
plt.xlabel("Subscribed to Term Deposit")
plt.ylabel("Age")
plt.title("Age vs Subscription")
plt.show()
```

```python
#TASK 4-- Create a bar plot using Seaborn to show the percentage of customers who subscribed to a term deposit for each education level (column 'education').<br><br><em><strong>Hint:</strong> Calculate the percentage for each group before plotting.</em>
pct_by_education = bank_df.groupby("education")["y"].apply(lambda x: (x == "yes").mean() * 100).reset_index(name="pct")

sns.barplot(x="education", y="pct", data=pct_by_education)
plt.xlabel("Education")
plt.ylabel("Subscribed (%)")
plt.title("Subscription % by Education")
plt.xticks(rotation=45)
plt.show()
```

```python
#TASK 5--4 Use ChatGPT or Copilot to suggest two additional visualizations that could help a fintech startup like Cred understand which users are most likely to buy a new savings product, then implement one of those visualizations using Seaborn.
pct_by_job = bank_df.groupby("job")["y"].apply(lambda x: (x == "yes").mean() * 100).reset_index(name="pct")

sns.barplot(x="pct", y="job", data=pct_by_job)
plt.xlabel("Subscribed (%)")
plt.ylabel("Job")
plt.title("Subscription % by Job")
plt.show()
```
