task 1

Use Seaborn's regplot to visualize the relationship between the number of likes and number of comments on a set of Instagram posts (create a small DataFrame with at least 10 sample data points).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "likes": [
        500, 800, 1200, 1500, 2000,
        2500, 3000, 3500, 4200, 5000
    ],

    "comments": [
        20, 35, 45, 55, 70,
        85, 100, 120, 145, 170
    ]
}

df = pd.DataFrame(data)
df.head()
```

```python
sns.regplot(
    x="likes",
    y="comments",
    data=df
)

plt.xlabel("Number of Likes")
plt.ylabel("Number of Comments")
plt.title("Instagram Likes vs Comments")

plt.show()
```

Create a lmplot in Seaborn to show the regression line between food order price and delivery time using a sample Zomato-style dataset (minimum 15 rows with random values for price and delivery time).

task 3

Apply lowess smoothing to a regplot that visualizes the relationship between the number of songs in a playlist and total playlist duration (in minutes) for Spotify users.<br><br><em><strong>Hint:</strong> Use the lowess=True parameter in regplot to see the difference in fit.</em>

```python
data = {
    "songs": [
        5, 10, 15, 20, 25,
        30, 35, 40, 45, 50,
        55, 60, 65, 70, 75
    ],

    "duration": [
        18, 35, 48, 65, 80,
        95, 115, 125, 145, 155,
        180, 195, 210, 225, 245
    ]
}

df = pd.DataFrame(data)

sns.regplot(
    x="songs",
    y="duration",
    data=df
)

plt.xlabel("Number of Songs")
plt.ylabel("Playlist Duration (Minutes)")
plt.title("Songs vs Playlist Duration - Linear Regression")

plt.show()
```

LOWER

```python
sns.regplot(
    x="songs",
    y="duration",
    data=df,
    lowess=True
)

plt.xlabel("Number of Songs")
plt.ylabel("Playlist Duration (Minutes)")
plt.title("Songs vs Playlist Duration - LOWESS Smoothing")

plt.show()
```

Generate a faceted regression plot (using lmplot with col parameter) to compare the relationship between hours spent on YouTube and exam scores, split by 'preparation method' (e.g., 'self-study' vs 'coaching') in your DataFrame

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "youtube_hours": [
        1, 2, 3, 4, 5,
        6, 7, 8, 2, 3,
        4, 5, 6, 7, 8,
        9, 10, 3, 5, 7, 9
    ],

    "exam_score": [
        92, 88, 85, 82, 78,
        75, 72, 68, 90, 87,
        84, 80, 76, 73, 69,
        65, 62, 89, 81, 74, 67
    ],

    "preparation_method": [
        "self-study", "self-study", "self-study",
        "self-study", "self-study", "self-study",
        "self-study", "self-study", "self-study",
        "self-study",

        "coaching", "coaching", "coaching", "coaching",
        "coaching", "coaching", "coaching", "coaching",
        "coaching", "coaching", "coaching"
    ]
}

df = pd.DataFrame(data)

sns.lmplot(
    x="youtube_hours",
    y="exam_score",
    col="preparation_method",
    data=df
)

plt.show()
```

Modify a given lmplot that shows regression between Flipkart product price and number of reviews. Add a title, axis labels, and customize the marker color.<br><br><em><strong>Constraint:</strong> Do not change the regression logic—only update plot aesthetics.</em>

```python
data = {
    "price": [
        5000, 8000, 10000, 12000, 15000,
        18000, 22000, 25000, 30000, 35000
    ],

    "reviews": [
        100, 180, 250, 300, 450,
        520, 700, 850, 1000, 1200
    ]
}

df = pd.DataFrame(data)
```

```python
sns.lmplot(
    x="price",
    y="reviews",
    data=df,
    scatter_kws={"color": "green"}
)

plt.xlabel("Product Price (₹)")
plt.ylabel("Number of Reviews")
plt.title("Flipkart Product Price vs Number of Reviews")

plt.show()
```
