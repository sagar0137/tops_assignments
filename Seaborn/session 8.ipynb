task 1

Use sns.scatterplot() to visualize the relationship between the number of followers and the number of posts for 20 fictional Instagram users. Label the axes appropriately.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "user": [
        "User1", "User2", "User3", "User4", "User5",
        "User6", "User7", "User8", "User9", "User10",
        "User11", "User12", "User13", "User14", "User15",
        "User16", "User17", "User18", "User19", "User20"
    ],

    "followers": [
        500, 1200, 2500, 1800, 3500,
        5000, 4200, 7500, 6200, 9000,
        11000, 8500, 13000, 15000, 18000,
        20000, 22000, 25000, 28000, 30000
    ],

    "posts": [
        20, 35, 45, 30, 60,
        75, 55, 90, 80, 110,
        130, 100, 145, 160, 175,
        190, 210, 230, 250, 280
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
sns.scatterplot(
    x="followers",
    y="posts",
    data=df
)

plt.xlabel("Number of Followers")
plt.ylabel("Number of Posts")
plt.title("Instagram Followers vs Number of Posts")

plt.show()
```

task 2

Create a line plot using sns.lineplot() to show the daily number of food orders on Zomato for each day over a month. Use a list of 30 random order counts to simulate the data.

```python
import numpy as np 
np.random.seed(42)

days = list(range(1, 31))

orders = np.random.randint(100, 500, 30)

df = pd.DataFrame({
    "day": days,
    "orders": orders
})

df.head()
```

```python
sns.lineplot(
    x="day",
    y="orders",
    data=df,
    marker="o"
)

plt.xlabel("Day of Month")
plt.ylabel("Number of Orders")
plt.title("Daily Zomato Orders - One Month")

plt.show()
```

task 3

Using sns.relplot(), plot the relationship between movie ratings and the number of reviews for different genres (Action, Comedy, Drama) based on a sample dataset of 15 movies. Use the 'hue' parameter to differentiate genres.<br><br><em><strong>Hint:</strong> Create a DataFrame with columns: 'movie', 'rating', 'reviews', and 'genre'.</em>

```python

data = {
    "movie": [
        "Movie A", "Movie B", "Movie C", "Movie D", "Movie E",
        "Movie F", "Movie G", "Movie H", "Movie I", "Movie J",
        "Movie K", "Movie L", "Movie M", "Movie N", "Movie O"
    ],

    "rating": [
        4.2, 4.5, 3.9, 4.1, 4.6,
        3.8, 4.0, 4.3, 3.7, 4.2,
        4.5, 4.1, 3.9, 4.4, 4.7
    ],

    "reviews": [
        1200, 2500, 900, 1800, 3200,
        700, 1500, 2100, 600, 1300,
        2800, 1700, 1000, 2400, 3500
    ],

    "genre": [
        "Action", "Action", "Action", "Action", "Action",
        "Comedy", "Comedy", "Comedy", "Comedy", "Comedy",
        "Drama", "Drama", "Drama", "Drama", "Drama"
    ]
}

df = pd.DataFrame(data)

df.head()
```

```python
sns.relplot(
    x="rating",
    y="reviews",
    data=df,
    hue="genre"
)

plt.xlabel("Movie Rating")
plt.ylabel("Number of Reviews")
plt.title("Movie Ratings vs Reviews by Genre")

plt.show()
```

Generate a regression plot using sns.regplot() to analyze the correlation between the number of hours spent on YouTube and exam scores for a group of 25 students. Interpret whether the trend is positive or negative.
