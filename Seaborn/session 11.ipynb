task 1

Create a Seaborn barplot showing the number of orders per food category (e.g., Pizza, Burger, Biryani, Momos, Ice Cream) and save the chart as a PNG file with 300 DPI.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "category": [
        "Pizza",
        "Burger",
        "Biryani",
        "Momos",
        "Ice Cream"
    ],
    
    "orders": [
        450,
        380,
        520,
        300,
        250
    ]
}

df = pd.DataFrame(data)

sns.barplot(
    x="category",
    y="orders",
    data=df
)

plt.xlabel("Food Category")
plt.ylabel("Number of Orders")
plt.title("Orders by Food Category")

plt.show()

plt.savefig(
    "food_category_orders.png",
    dpi=300,
    bbox_inches="tight"
)
```

Generate a Seaborn line plot visualizing daily step counts for one week, then export the chart as an SVG file suitable for use in a PowerPoint presentation.<br><br><em><strong>Hint:</strong> Use plt.savefig('filename.svg', format='svg').</em>

```python
data = {
    "day": [
        "Mon", "Tue", "Wed",
        "Thu", "Fri", "Sat", "Sun"
    ],
    
    "steps": [
        6500, 7200, 8000,
        7500, 9000, 10500, 8500
    ]
}

df = pd.DataFrame(data)

sns.lineplot(
    x="day",
    y="steps",
    data=df,
    marker="o"
)

plt.xlabel("Day")
plt.ylabel("Daily Steps")
plt.title("Weekly Daily Step Counts")

plt.savefig(
    "weekly_steps.svg",
    format="svg",
    bbox_inches="tight"
)

plt.show()
```

task 4

Create a template function called apply_presentation_theme() that applies a consistent Seaborn style, font, and color palette to any chart, then use it to style and export a pie chart showing your top 5 most-used apps

```python
import seaborn as sns
import matplotlib.pyplot as plt

def apply_presentation_theme():
    sns.set_theme(
        style="whitegrid",
        context="talk",
        palette="deep",
        font_scale=1.0
    )

apply_presentation_theme()

import pandas as pd

data = {
    "app": [
        "Instagram",
        "YouTube",
        "WhatsApp",
        "Spotify",
        "Chrome"
    ],
    
    "usage_hours": [
        20,
        18,
        15,
        12,
        10
    ]
}

df = pd.DataFrame(data)
plt.figure(figsize=(8, 8))

plt.pie(
    df["usage_hours"],
    labels=df["app"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Most-Used Apps")

plt.savefig(
    "top_5_apps.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
data = {
    "sales": [100, 150, 200, 250, 300],
    "quantity": [10, 15, 20, 25, 30],
    "discount": [5, 8, 10, 12, 15],
    "profit": [20, 35, 45, 55, 70]
}

df = pd.DataFrame(data)

corr = df.corr()



plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Sales Correlation Heatmap")

plt.savefig(
    "sales_correlation_1200dpi.svg",
    format="svg",
    dpi=1200,
    bbox_inches="tight"
)

plt.show()
```
